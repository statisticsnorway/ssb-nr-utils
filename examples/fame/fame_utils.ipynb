# FAME utils
Lagde disse utils funksjonene for å bedre bruke FAME fra python. Totalt tre funksjoner

* create_fame_db: lager en ny famedb. KAn også brukes til å overskrive en fame db med data.
* get_fame: Lager en csv fil fra uttrekk fra fame db. Henter alle variabler, men du kan begrense datoene.
* update_fame_db: Oppdaterer fame db. Denne funksjonen både appenderer nye data og oppdaterer tidligere datoer, dersom det finnes data i pandas df.

```python
import pandas as pd
from nr_utils.fame.fame import *

navn = input()
home_path = f"/ssb/bruker/{navn}"

print("bruker filsti:", home_path)

start = "2026-01-01"
end = "2026-03-01"
```

## create_fame_db
Her ser du at jeg oppretter en ny fame db med to kolonner og tre perioder. 

```python
df = pd.DataFrame(
    data={"test.1.vr": [900, 1000, 1100], "test.2.vl": [9, 10, 11]},
    index=pd.date_range(start="2026-01-01", end="2026-03-01", freq="MS"),
)

create_fame_db(
    data=df,
    freq="m",
    start_date="2026:1",
    end_date="2026:3",
    db_path=f"{home_path}/my_first_db.db",
    db_alias="mydb",
)
```

## get_fame
La oss hente den samme fame dben med get_fame. Da kan vi se på dataene som en csv fil.

```python
get_fame(
    db_path="/ssb/bruker/ged/my_first_db.db",
    csv_path=f"{home_path}/test_csv.csv",
    freq="m",
    start_date="2026:1",
    end_date="2026:3",
    rounding="auto",
    famedb="sl-fame-p1",
)

pd.read_csv(f"{home_path}/test_csv.csv", sep=";")
```

## update_fame_db
La oss oppdatere fame med ny data. Her har vi data for mar til mai 2026. Da vil to ting skje. Mars 2026 oppdateres og april og mai appenderes inn.

```python
df = pd.DataFrame(
    data={"test.1.vr": [9999, 8888, 7777], "test.2.vl": [9, 10, 11]},
    index=pd.date_range(start="2026-03-01", end="2026-05-01", freq="MS"),
)

update_fame_db(
    data=df,
    freq="monthly",
    start_date="2026:3",
    end_date="2026:5",
    db_path=f"{home_path}/my_first_db.db",
    famedb="sl-fame-p1",
)
```
